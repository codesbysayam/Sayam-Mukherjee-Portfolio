```python
import os
import json
import urllib.request
from collections import Counter

# Locate local repository root
repo_root = os.path.abspath(os.path.join(os.getcwd(), "../.."))
data_store_path = os.path.join(repo_root, "data_store.json")
print(f"Target repository root: {repo_root}")
print(f"Data store exists: {os.path.isfile(data_store_path)}")
```

## 1. Project & Technology Relationship Analysis
Inspect verified projects and map out technology distributions across projects.

```python
with open(data_store_path, "r", encoding="utf-8") as f:
    data = json.load(f)

projects = data.get("projects", [])
print(f"Total verified projects loaded: {len(projects)}")

tech_counts = Counter()
for p in projects:
    for tag in p.get("tags", []):
        tech_counts[tag] += 1

print("\n--- Technology Frequency Across Projects ---")
for tech, count in tech_counts.most_common():
    print(f"{tech:<25} | {count} projects")
```

## 2. GitHub Codebase Language Byte Analysis
Query public GitHub language byte counts for the portfolio repository and calculate accurate composition percentages.

```python
REPO_NAME = "codesbysayam/Sayam-Mukherjee-Portfolio"
api_url = f"https://api.github.com/repos/{REPO_NAME}/languages"

# Baseline verified bytes snapshot if offline or rate-limited
language_bytes = {
    "TypeScript": 846240,
    "CSS": 36875,
    "JavaScript": 9649,
    "HTML": 2016,
    "Python": 12500,
    "C": 4200
}

try:
    req = urllib.request.Request(api_url, headers={"User-Agent": "Python-Portfolio-Analysis"})
    with urllib.request.urlopen(req, timeout=5) as response:
        if response.status == 200:
            live_data = json.loads(response.read().decode("utf-8"))
            if live_data:
                language_bytes = live_data
                print("Successfully fetched live data from GitHub API.")
except Exception as e:
    print(f"GitHub API query note ({e}); using local verified baseline snapshot.")

# Calculate exact percentage distribution
total_bytes = sum(language_bytes.values())
distribution = [
    {
        "language": lang,
        "bytes": count,
        "percentage": round((count / total_bytes) * 100, 2)
    }
    for lang, count in language_bytes.items()
]
distribution.sort(key=lambda x: x["bytes"], reverse=True)

print(f"\nTotal Codebase Bytes: {total_bytes:,} bytes ({(total_bytes / 1024):,.1f} KB)")
print("-" * 55)
print(f"{'Language':<18} | {'Bytes':<12} | {'Share (%)':<10}")
print("-" * 55)
for item in distribution:
    bar = "█" * int(item["percentage"] / 2.5)
    print(f"{item['language']:<18} | {item['bytes']:<12,} | {item['percentage']:>5.1f}%  {bar}")
```
